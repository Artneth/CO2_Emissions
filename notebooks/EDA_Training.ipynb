### Before Standardization - Histplot

```python
plt.figure(figsize=(18, 6))
for i, column in enumerate(numerical_col, 1):
    plt.subplot(1, len(numerical_col), i)
    sns.histplot(data[column], kde=True, bins=20)
    plt.title(f'Before Standardization: {column}')
plt.tight_layout()
plt.show()
```

## Standardize the numerical columns using Z-score

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_col = data.select_dtypes(include=np.number).columns

scaler = StandardScaler()
data1 = data.copy()
data1[numerical_col] = scaler.fit_transform(data[numerical_col])
```

### After Standaardization - Histplot

```python
plt.figure(figsize=(18, 6))
for i, column in enumerate(numerical_col, 1):
    plt.subplot(1, len(numerical_col), i)
    sns.histplot(data1[column], kde=True, bins=20)
    plt.title(f'After Standardization: {column}')
plt.tight_layout()
plt.show()
```

```python
for column in numerical_col:
    original_mean = data[column].mean()
    original_std = data[column].std()
    standardized_mean = data1[column].mean()
    standardized_std = data1[column].std()
    print(f"\n{column}:")
    print(f"  Before Standardization -> Mean: {original_mean}, Std Dev: {original_std}")
    print(f"  After Standardization -> Mean: {standardized_mean}, Std Dev: {standardized_std}")
```

```python
#DROP THE UNNESSCARY COLUMNS
data1=data1.drop(columns=['make','model'])
```

```python
#ENCODE THE CATEGORIAL VARIABLES
categorical_cols = ["vehicle_class", "transmission", "fuel_type"]

data1 = pd.get_dummies(data1, columns=categorical_cols, drop_first=True, dtype=int)
```

```python
data1.head()
```

#Train/Test split

```python
X =data1.drop(columns=["co2_emissions"])
y = data1["co2_emissions"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
print(f"Training Data Shape: {X_train.shape}")
print(f"Testing Data Shape: {X_test.shape}")
print(f"Number of Features: {X_train.shape[1]}")
```

# Train Different Types of Models

## linear regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# make prediction
y_pred = model.predict(X_test)
```

```python
#  Evaluate the Model
mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
```

```python
print("Model Performance:")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```

```python

threshold = y_test.mean()
y_pred_class = (y_pred >= threshold).astype(int)
y_test_class = (y_test >= threshold).astype(int)

acc_r = accuracy_score(y_test_class, y_pred_class) * 100
print("Accuracy =", acc_r)
```

```python
accuracy_linear=model.score(X_test,y_test)
print("Accuracy_linear:",accuracy_linear)
```

```python
acc1 = model.score(X_train, y_train)
print("Accuracy =", acc1)
```

## Kmeans Clustering


```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score, davies_bouldin_score
```

```python
X_scaled = X
```

```python
inertia = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=1, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
# Plot the Elbow Method
plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker='o', linestyle='--')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")
plt.show()
```

```python
optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=1, n_init=10)
clusters = kmeans.fit_predict(X_scaled)
```

```python
data1["Cluster"] = clusters
```

```python
inertia = kmeans.inertia_
silhouette = silhouette_score(X_scaled, clusters)
davies_bouldin = davies_bouldin_score(X_scaled, clusters)
```

```python
print(f"🔹 K-Means Clustering Evaluation for {optimal_k} Clusters:")
print(f"Inertia (WCSS): {inertia:.2f} (Lower is better)")
print(f"Silhouette Score: {silhouette:.2f} (Higher is better, range -1 to 1)")
print(f"Davies-Bouldin Index: {davies_bouldin:.2f} (Lower is better)")
```

```python
data1['Cluster'] = kmeans.labels_

# Plotting the scatter plot
plt.figure(figsize=(10, 7))
sns.scatterplot(x='engine_size', y='co2_emissions', hue='Cluster', data=data1, palette='viridis', s=100)
plt.title('K-Means Clustering Results')
plt.xlabel('Engine Size')
plt.ylabel('CO2 Emissions')
plt.show()
```
